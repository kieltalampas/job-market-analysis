### DATA CLEANING

```python
print(df.columns.tolist())
```

```python
columns_to_keep = [
    "job_id",
    "search_role",
    "job_title",
    "employer_name",
    "job_publisher",
    "job_employment_type",
    "job_description",
    "job_is_remote",
    "job_posted_at_datetime_utc",
    "job_location",
    "job_city",
    "job_state",
    "job_country"
]

df_clean = df[columns_to_keep].copy()
```

```python
df_clean["job_posted_at_datetime_utc"] = pd.to_datetime(
    df_clean["job_posted_at_datetime_utc"],
    errors="coerce"
)
```

```python
print(df_clean["job_employment_type"].value_counts(dropna=False))
```

```python
print(df_clean["job_is_remote"].value_counts(dropna=False))
```

```python
df_clean["location_clean"] = df_clean["job_location"]

df_clean["location_clean"] = df_clean["location_clean"].fillna(
    df_clean["job_city"]
)

df_clean["location_clean"] = df_clean["location_clean"].fillna(
    df_clean["job_state"]
)
```

```python
df_clean.loc[
    df_clean["location_clean"].isna(),
    [
        "search_role",
        "job_title",
        "employer_name",
        "job_location",
        "job_city",
        "job_state",
        "job_country"
    ]
]
```

```python
df_clean.loc[556, "location_clean"] = "Cebu, Philippines"
```

```python
df_clean.loc[235, "location_clean"] = "Makati, Philippines"
```

```python
print(df_clean.info())
```

```python
print(df_clean["job_employment_type"].value_counts(dropna=False))
```

```python
df_clean["job_employment_type"] = (
    df_clean["job_employment_type"]
    .fillna("Unknown")
    .str.strip()
)
```

```python
columns_to_drop = [
    "job_min_salary",
    "job_max_salary",
    "job_salary",
    "job_salary_string",
    "job_salary_period",
    "job_benefits",
    "job_benefits_strings",
    "job_onet_soc",
    "job_onet_job_zone",
    "employer_reviews"
]

df_clean = df_clean.drop(columns=columns_to_drop, errors="ignore")
```

# SKILL EXTRACTION

```python
skills = [
    "Python",
    "SQL",
    "Power BI",
    "Tableau",
    "Excel",
    "AWS",
    "Azure",
    "GCP",
    "Spark",
    "PySpark",
    "Snowflake",
    "BigQuery",
    "Databricks",
    "Airflow",
    "dbt",
    "Looker",
    "Git",
    "Docker",
    "Kubernetes",

    # Software engineering
    "Java",
    "JavaScript",
    "TypeScript",
    "C#",
    "C++",
    "React",
    "Angular",
    "Node.js",

    # Databases / infrastructure
    "PostgreSQL",
    "MySQL",
    "MongoDB",
    "Oracle",
    "Linux",
    "Terraform",

    # Data science / ML
    "Machine Learning",
    "TensorFlow",
    "PyTorch",
    "Scikit-learn"
]
```

```python
skill_rows = []

for _, row in df_clean.iterrows():

    text = (
        str(row["job_title"]) + " " +
        str(row["job_description"])
    ).lower()

    for skill in skills:

        if skill.lower() in text:

            skill_rows.append({
                "job_id": row["job_id"],
                "search_role": row["search_role"],
                "skill": skill
            })
```

```python
df_skills = pd.DataFrame(skill_rows)
```

```python
print(
    df_skills.duplicated(
        subset=["job_id", "skill"]
    ).sum()
)

print(df_skills["job_id"].nunique())

skill_coverage = (
    df_skills["job_id"].nunique()
    / df_clean["job_id"].nunique()
    * 100
)

print(skill_coverage)
```

```python
matched_job_ids = df_skills["job_id"].unique()

unmatched_jobs = df_clean[
    ~df_clean["job_id"].isin(matched_job_ids)
]

print(unmatched_jobs.shape)

unmatched_jobs[
    [
        "search_role",
        "job_title",
        "job_description"
    ]
].head(30)
```
